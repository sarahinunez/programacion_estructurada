```python
MAX_INTENTOS = 3
PIN_CORRECTO_PREDETERMINADO = 1219


def verificar_pin():
    print("Configure el PIN correcto (deje vacío para usar 1219):")
    try:
        entrada_inicial = input()
        if entrada_inicial.strip():
            pin_correcto = int(entrada_inicial)
        else:
            pin_correcto = PIN_CORRECTO_PREDETERMINADO
    except (ValueError, EOFError):
        pin_correcto = PIN_CORRECTO_PREDETERMINADO

    print(f"PIN correcto establecido. Tiene {MAX_INTENTOS} intentos para ingresar su PIN.")
    intentos = 0
    mensaje = ""

    while True:
        try:
            print(f"Intento {intentos + 1}: Ingrese su PIN:")
            entrada_usuario = input()
            pin_tecleado = int(entrada_usuario)
        except ValueError:
            print("Escribe un número entero.")
            continue
        except EOFError:
            mensaje = "DENEGADO"
            break

        intentos += 1

        if pin_tecleado == pin_correcto:
            mensaje = "ACCESO"
            break
        else:
            if intentos == MAX_INTENTOS:
                mensaje = "DENEGADO"
                break

    print(f"\nResultado: {mensaje}")


if __name__ == "__main__":
    verificar_pin()
```
