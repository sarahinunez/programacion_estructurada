```python
import numpy as np

sistolica = np.array([118, 135, 142, 110, 128, 155, 121])
promedio = np.mean(sistolica)
desviacion = np.std(sistolica)

conteo = np.sum(sistolica >= 130)
posiciones = np.nonzero(sistolica >= 130)

print("Promedio: ", promedio)
print("Desviacion estandar: ", desviacion)
print("Conteo >= 130: ", conteo)
print("Posiciones:", posiciones)
```

```python
import numpy as np

glucosa = np.array([[95, 102, 99],
                   [140, 155, 148],
                   [88, 91, 85],
                   [110, 118, 125],
                   ])

promedio_pacientes = np.mean(glucosa, axis=1)
prom_dia = np.mean(glucosa, axis=0)
dia_max = np.argmax(prom_dia)

indice_plano_max = np.argmax(glucosa)
paciente_max, dia_max_pos = np.unravel_index(indice_plano_max, glucosa.shape)

print("promedio por paciente:", promedio_pacientes)
print("promedio por dia:", prom_dia)
print("dia con el promedio mas alto (indice):", dia_max)
print(f"lectura mas alta registrada: {glucosa[paciente_max, dia_max_pos]}")
print(f"paciente con el maximo (indice): {paciente_max}")
print(f"dia con el maximo (indice): {dia_max_pos}")
```
