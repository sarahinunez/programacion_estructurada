```python
import pandas as pd
from google.colab import files

df_pacientes = pd.read_csv('/content/pacientes (1).csv')
pacientes_lista = df_pacientes.dropna(subset=['prioridad', 'apellido', 'fecha_ingreso']).to_dict('records')

prioridad_map = {'ROJO': 1, 'AMARILLO': 2, 'VERDE': 3}

def obtener_prioridad_val(paciente):
    prio = str(paciente.get('prioridad', '')).upper()
    return prioridad_map.get(prio, 99)

def es_mayor(pac_act, pac_sig):
    val_prio_act = obtener_prioridad_val(pac_act)
    val_prio_sig = obtener_prioridad_val(pac_sig)
    if val_prio_act > val_prio_sig:
        return True
    elif val_prio_act == val_prio_sig:
        ape_act = str(pac_act.get('apellido', '')).lower()
        ape_sig = str(pac_sig.get('apellido', '')).lower()
        if ape_act > ape_sig:
            return True
        elif ape_act == ape_sig:
            f_act = str(pac_act.get('fecha_ingreso', ''))
            f_sig = str(pac_sig.get('fecha_ingreso', ''))
            if f_act > f_sig:
                return True
    return False

def merge_sort(datos):
    if len(datos) <= 1:
        return datos

    mitad = len(datos) // 2
    izquierda = merge_sort(datos[:mitad])
    derecha = merge_sort(datos[mitad:])

    return mezclar(izquierda, derecha)

def mezclar(izquierda, derecha):
    resultado = []
    i = 0
    j = 0

    while i < len(izquierda) and j < len(derecha):
        if es_mayor(izquierda[i], derecha[j]):
            resultado.append(derecha[j])
            j += 1
        else:
            resultado.append(izquierda[i])
            i += 1

    resultado.extend(izquierda[i:])
    resultado.extend(derecha[j:])
    return resultado

pacientes_ordenados = merge_sort(pacientes_lista)
df_ordenado = pd.DataFrame(pacientes_ordenados)

print(f"\nTotal de pacientes ordenados: {len(df_ordenado)}")
pd.set_option('display.max_rows', 20)
display(df_ordenado[['prioridad', 'apellido', 'nombre', 'fecha_ingreso']])

df_ordenado.to_csv('/content/pacientes_ordenados.csv', index=False)

try:
    files.download('/content/pacientes_ordenados.csv')
    print("Descarga de 'pacientes_ordenados.csv' iniciada con éxito.")
except Exception as e:
    print(f"No se pudo iniciar la descarga automática: {e}")
```
